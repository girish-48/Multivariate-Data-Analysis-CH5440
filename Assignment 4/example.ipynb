```python
Concentrations = mat['CONC']
Absorbance     = mat['DATA']
```

```python
A = Absorbance[0::5,[47,53,105]]
C = Concentrations[0::5]

A = A - A.mean(axis=0)
C = C - C.mean(axis=0)
```

```python
B = ((A.T@A)**-1)@A.T@C
```
