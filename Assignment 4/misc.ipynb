```python
from scipy.signal import argrelextrema
import numpy as np
```

```python
array = np.array([ 5,4,3,2,1.5,1.5,2,3,0,0,0])

print(argrelextrema(array, np.greater))
# returns the local maximas of a discrete function 
print(argrelextrema(array, np.less))
# returns the local minimas of a discrete function 
```
