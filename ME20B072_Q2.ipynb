```python
vals = os.listdir(r'yalefaces')
vals.sort()
```

```python
names = np.array(vals).reshape((15,6))

data = np.empty((77760,1))
for name in names :
    for image_name in name :
        print(image_name)
        filename = 'yalefaces/' + image_name
        img = Image.open(filename)
        img_vec = np.array(img).T.reshape((77760,1))
        np.append(data, img_vec, axis=1)
        plt.imshow(img)
        img.close()
```

```python
with Image.open(filename) as img : 
            img.load()
            img_vec = np.array(img).T.reshape((77760,1))
            np.append(data, img_vec, axis=1)
            plt.imshow(img)
```
