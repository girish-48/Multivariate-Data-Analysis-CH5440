```python
import numpy as np
import numpy.linalg as linal
import scipy.io 
import matplotlib.pyplot as plt
mean_shift = lambda x : x - x.mean(axis=0)
```

```python
data = scipy.io.loadmat('reactiondata.mat')
```

```python
data
```

```python
np.shape(data['temp'])
```

```python
Ca = data['Ca']
ra = -(Ca[1:] - Ca[:-1])
```

```python
u = np.log(Ca)
um = mean_shift(u)
y = np.log(ra)
ym = mean_shift(y)
```

```python
from sklearn import linear_model
```

```python
beta = []
beta0 = []
for i in range(6) : 
    ua = u[:-1,i] [np.newaxis] .T 
    ya = y[:,i] [np.newaxis] .T 
    reg = linear_model.LinearRegression().fit(ua,ya)
    beta.append(reg.coef_[0,0])
    beta0.append(reg.intercept_[0])
```

```python
beta # n values for different temperatures 
```

```python
k = np.exp(beta0) #  k values 
```

```python
T = data['temp']
x = -1/T
y1 = beta0
```

```python
reg1 = linear_model.LinearRegression().fit(x,y1)
```

```python
log_k0 = reg1.intercept_
E      = reg1.coef_
k0     = np.exp(log_k0)
```

```python
print("Equation for K : ")
print(f"{np.round(k0,3)} * exp( -{np.round(E[0],3)}/T )")
# the is only a single value of k0 and E for this entire experiment so the mean remains the same value as the original value 
```

```python
print(f"Estimated values of rate constants : ")
for i in range(6) : 
    print(f"k = {np.round(k[i],5)} \t:\t T = {np.round(T[i,0],3)} K")
```

```python
print(f"Mean value of reaction order = {np.round(np.mean(beta),4)}")
```
