```python
import scipy.io
from scipy.stats.distributions import chi2
import scipy.linalg as slinal
import numpy as np 
import numpy.linalg as linal
import pandas as pd 
import sympy as sp
```

```python
mat = scipy.io.loadmat('cov.mat')
COV = np.matrix(mat['covF'])
```

```python
eigenvalues, eigenvectors = linal.eig(COV)

print("EIGENVALUES AND EIGENVECTORS :")
print("-----------------------------------------------------------------------------------------------------------")

for eig in eigenvalues : 
    index = np.where(eigenvalues == eig)[0][0] 
    print (f"{index + 1} |", np.round(eig,5), end='\t')
    print ("| Eigenvector :", *eigenvectors[:,index].T, sep='\t')

print("-----------------------------------------------------------------------------------------------------------")
```

$$ Az^{*}_{i} = 0 $$
* $z^{*}_{i} : \textbf{True value of measurements of i th sample}$ 
* $A : \textbf{mxp constraint matrix} $ 
* $p : \textbf{Number of Variables} $ 
* $N : \textbf{Number of Samples in data Matrix} $ 
* $m : \textbf{Number of Contraints}$ 

$$ H_0 : \lambda_{N,p-d+1} = \lambda_{N,p-d+2} = ... = \lambda_{N,p} $$
$$ H_1 : \textbf{at least two smallest d eigenvalues are unequal} $$

* The Statistic to test the Hypothesis is : 
$$ \tau = n' \cdot \left[ d\cdot \ln{\overline{l}} - \sum_{j = p-d+1}^{j = p} \ln{l_{N,j}}\right] $$
$$ \overline{l} = \frac{1}{d}\cdot \sum_{j = p-d+1}^{j = p} l_{N,j}$$
$$ n' = N - \frac{2p+11}{6} $$

* Here $l_{N,j}$ is the jth eigenvalue of $S_{z_s}$
* The NULL HYPOTHESIS is rejected if the $\tau \ge \chi^2_{\nu,\alpha}$
* Here $\chi^2$ is the chi-square distirbution with $\nu$ degrees of freedom, computed as $\nu = \frac{1}{2}\cdot (d+2)\cdot (d-1)$


```python
p = 6
d = p-1
N = 1000

print("STATISTICAL TESTING :")
while True : 
    n_dash = N - (2*p+11)/6
    l_dash = eigenvalues[ p-d : ].sum()/d 

    tau    = n_dash * ( d * np.log(l_dash) - np.log(eigenvalues[ p-d : ]).sum())

    fdom   = 0.5*(d+2)*(d-1)
    chi    = chi2.ppf(0.95, df=fdom)

    print("-----------------------------------------------------")
    print(f"Chi-square at 95% and {int(fdom)} dof \t\t: {np.round(chi,4)}")
    print(f"The computed statistic value is \t: {np.round(tau,4)}")

    if (tau <= chi) : 
        break 
    else : 
        d = d - 1 

print("-----------------------------------------------------",end='\n\n')
print (f"Number of constraints after performing the above test is : {d}")
```

```python
print("EIGENVALUES AND EIGENVECTORS AFTER PERFORMING STATISTIC TESTING:")
print("-----------------------------------------------------------------------------------------------------------")

for eig in eigenvalues[d:] : 
    index = np.where(eigenvalues == eig)[0][0] 
    print (f"{index + 1} |", np.round(eig,5), end='\t')
    print ("| Eigenvector :", *eigenvectors[:,index].T, sep='\t')

print("-----------------------------------------------------------------------------------------------------------")
```

* The correct number of linearly independent constraints have been determined as the system has only 3 linearly independent constraints and it is the same result we obtained by performing Statistic test
